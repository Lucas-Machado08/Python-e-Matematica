```python
import pandas as pd
import plotly.express as px

# 1. Carregamento e Higienização dos Dados
caminho_ficheiro = 'Tabela_13_Meio_de_transporte.xlsx'

# Leitura da aba do Brasil com divisão por Cor/Raça
try:
    df_brasil = pd.read_excel(caminho_ficheiro, sheet_name='Exemplo gráfico Brasil', skiprows=1)
    df_brasil.columns = ['Meio de Transporte', 'Branca', 'Preta ou Parda', 'Indígena']

    # 2. Processamento para Mediana por Meio de Transporte por Raça (KPIs)
    df_melted = df_brasil.melt(
        id_vars=['Meio de Transporte'], 
        var_name='Raça/Cor', 
        value_name='Percentual'
    )

    # Mediana por meio de transporte e raça
    mediana_modal_raca = df_melted.groupby(['Meio de Transporte', 'Raça/Cor'])['Percentual'].median().reset_index()
    print('--- Mediana por Meio de Transporte e Raça ---')
    print(mediana_modal_raca)

    # 3. Análise de Modal Público vs. Particular vs. Ativo
    def categorizar_modal(modal):
        modal_str = str(modal).lower()
        if 'coletivo' in modal_str:
            return 'Público'
        elif 'automóvel' in modal_str or 'moto' in modal_str:
            return 'Particular'
        elif 'bicicleta' in modal_str or 'pé' in modal_str:
            return 'Ativo'
        else:
            return 'Outros'

    df_melted['Categoria Modal'] = df_melted['Meio de Transporte'].apply(categorizar_modal)
    uso_por_categoria = df_melted.groupby('Categoria Modal')['Percentual'].sum().reset_index()

    # 4. Exportação dos Gráficos do Projeto
    fig = px.bar(
        df_melted, 
        x='Meio de Transporte', 
        y='Percentual', 
        color='Raça/Cor', 
        barmode='group',
        title='Uso de Modais de Transporte por Raça/Cor no Brasil'
    )
    fig.write_html('grafico_interativo_transporte.html')
    print('Gráfico interativo gerado com sucesso!')
except Exception as e:
    print(f'Erro ao carregar os dados: {e}')
```
